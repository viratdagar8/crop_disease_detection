## 2. Load PlantVillage Dataset (Color Images)

```python
# Load the official color configuration from Hugging Face
dataset = load_dataset("mohanty/PlantVillage", "color")
print(dataset)

labels = dataset["train"].features["label"].names
print(f"Total disease classes: {len(labels)}")
```

## 3. Visualize Sample Leaf Images

```python
plt.figure(figsize=(15, 8))
for i in range(8):
    sample = dataset["train"][i * 500]
    img = sample["image"]
    lbl = labels[sample["label"]]
    
    plt.subplot(2, 4, i + 1)
    plt.imshow(img)
    plt.title(lbl.replace("___", "\n"), fontsize=9)
    plt.axis("off")

plt.tight_layout()
plt.show()
```
